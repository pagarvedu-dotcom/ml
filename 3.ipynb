```python


# Gradient Descent Algorithm

x = 2
lr = 0.1
iterations = 100

for i in range(iterations):
    gradient = 2 * (x + 3)
    x = x - lr * gradient

    if (i + 1) % 10 == 0:
        print("Iteration:", i + 1, "x:", round(x, 4))

y = (x + 3) ** 2

print("Minimum x:", round(x, 2))
print("Minimum y:", round(y, 2))
```

```python

hours = [1, 2, 3, 4, 5]
marks = [20, 30, 40, 50, 60]

w = 0
b = 0
lr = 0.01
n = len(hours)

for i in range(10000):
    dw = 0
    db = 0
    cost = 0

    for j in range(n):
        pred = w * hours[j] + b
        error = pred - marks[j]

        dw += error * hours[j]
        db += error
        cost += error ** 2

    dw = dw / n
    db = db / n
    cost = cost / (2 * n)

    w = w - lr * dw
    b = b - lr * db

print("Final Weight:", round(w, 2))
print("Final Bias:", round(b, 2))
print("Final Cost:", round(cost, 4))

new_hours = 6
predicted_marks = w * new_hours + b

print("Predicted Marks:", round(predicted_marks, 2))

```
