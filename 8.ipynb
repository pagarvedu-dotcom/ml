```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
from mlxtend.preprocessing import TransactionEncoder
```

```python
# Step 1: Create Dataset
data = [
    ['Bread', 'Butter', 'Jam', 'Juice', 'Milk'],
    ['Jam', 'Juice', 'Bread', 'Butter'],
    ['Juice', 'Milk', 'Bread'],
    ['Butter', 'Juice', 'Biscuit', 'Milk'],
    ['Bread', 'Milk', 'Butter']
]
```

```python
# Step 2: Convert Data
te = TransactionEncoder()
encoded = te.fit(data).transform(data)
df = pd.DataFrame(encoded, columns=te.columns_)
```

```python
print("Dataset:")
print(df)
```

```python
# Step 3: Apply Apriori
frequent_items = apriori(df, min_support=0.4, use_colnames=True)

print("Frequent Itemsets:")
print(frequent_items)
```

```python

# Step 4: Generate Rules
rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.6
)
```

```python
# Step 5: Display Results
print("Association Rules:")
print(rules[['antecedents', 'consequents',
             'support', 'confidence', 'lift']])
```

```python

import pandas as pd
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules

# Step 1: Load Dataset
df = pd.read_csv("Groceries.csv")
print(df.head())

# Step 2: Create Transactions
transactions = []

for i in range(len(df)):
    items = df.iloc[i].dropna().astype(str).tolist()
    transactions.append(items)

# Step 3: Encode Transactions
te = TransactionEncoder()
data = te.fit(transactions).transform(transactions)
df_encoded = pd.DataFrame(data, columns=te.columns_)

# Step 4: Apply Apriori
frequent_items = apriori(
    df_encoded, min_support=0.01, use_colnames=True
)

print("Frequent Itemsets:")
print(frequent_items)

# Step 5: Generate Association Rules
rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.3
)

# Step 6: Display Results
print("Association Rules:")
print(rules[['antecedents', 'consequents',
             'support', 'confidence', 'lift']])
```
