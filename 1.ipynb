```python
import pandas as pd
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler ,LabelEncoder
```

```python
df=pd.read_csv('Studen.csv')
```

```python
print(df)
```

```python
# Step 2: Check missing values
print('\nMissing Values:')
print(df.isnull().sum())
```

```python
# Step 3: Fill missing values
df[['Age','Attendance','Marks','Income']]=df[['Age','Attendance','Marks','Income']].fillna(df[['Age','Attendance','Marks','Income']].mean())
```

```python
df
```

```python
df.isnull().sum()
```

```python
# Step 4: Remove duplicate records
print(df.duplicated())
```

```python
df = df.drop_duplicates()
```

```python
# Step 5: Handle outliers using IQR
```

```python
Q1=df['Income'].quantile(0.25)
Q3=df['Income'].quantile(0.75)
```

```python
IQR =Q3-Q1
```

```python
lower= Q1-1.5 * IQR
upper= Q3+1.5 * IQR
```

```python
quantiles=df[(df['Income']> lower) |
             (df['Income']<upper)
            ]
             
```

```python
print("Q1:", Q1)
print("Q2:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
# Handle outliers
df['Income'] = df['Income'].clip(lower, upper)
```

```python
print(df['Income'])
```

```python
# Step 6: Label Encoding
df['Gender']=df['Gender'].map({'Female':0,'Male':1})
```

```python
df
```

```python
# Step 7: Normalization
features = ['Attendance']
```

```python
scaler = MinMaxScaler()
```

```python
df[features] = scaler.fit_transform(df[features])
```

```python
# Step 8: Standardization
features = ['Marks']
```

```python
scaler = StandardScaler()
```

```python
df[features] = scaler.fit_transform(df[features])
```
